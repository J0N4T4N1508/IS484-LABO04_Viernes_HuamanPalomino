## Representacion de un problema mediante estados

### Parte A: Completar la cha del problema para el siguiente caso: un repartidor debe ir desde el Almacén hasta la Zona Norte, pudiendo pasar por distintos puntos intermedios de la ciudad.

| Elemento | Descripcion|
|---|---|
|**Estado inicial** | Almacen |
|**Estado Objetivo** | Zona norte |
|**Acciones posibles** | Desplzarse desde un punto de la ciudad hacia otro punto conectado directamente en el mapa |
|**Costo de cada accion** | 1 paso por cada desplazamiento entre dos puntos conectados |

### Parte B: Representar el mapa de la ciudad como un grafo en Python, usando un diccionario de listas (cada clave es un punto, cada valor es la lista de puntos conectados directamente):

```python
mapa = {
"Almacen": ["Centro", "Terminal"],
"Centro": ["Almacen", "ZonaNorte", "Mercado"],
"Terminal": ["Almacen", "Mercado"],
"Mercado": ["Centro", "Terminal", "ZonaNorte"],
"ZonaNorte": ["Centro", "Mercado"],
}
print(mapa)
```

```python
def bfs(grafo, inicio, objetivo):
    """
    Busca el camino mas corto entre 'inicio' y 'objetivo' usando BFS.
    """
    cola = [[inicio]]  # cola de caminos, empieza con el estado inicial
    visitados = set()

    while cola:
        camino = cola.pop(0)  # sacar el primer camino de la cola (FIFO)
        nodo_actual = camino[-1]  # [-1] significa "último elemento de la lista".

        if nodo_actual == objetivo:
            # Devolvemos el camino completo.
            return camino

        if nodo_actual not in visitados:
            visitados.add(nodo_actual)

            for vecino in grafo[nodo_actual]:
                # Crear un nuevo camino: camino actual + vecino
                nuevo_camino = camino + [vecino]

                # Agregar el nuevo camino a la cola
                cola.append(nuevo_camino)

    return None  # no se encontro un camino


resultado = bfs(mapa, "Almacen", "ZonaNorte")
print("Camino encontrado:", resultado)
```

## Parte A: Implementación de DFS

**Completar la búsqueda en profundidad sobre el mismo mapa de la Sección 4.2, usando una pila
(lista de Python, con .pop() para sacar el último elemento):**

```python
def dfs(grafo, inicio, objetivo):
    """
    Busca un camino entre 'inicio' y 'objetivo' usando DFS.
    """

    # Creamos una pila que contiene inicialmente
    # el camino que comienza en el nodo inicial.
    pila = [[inicio]]

    # Conjunto donde guardaremos los nodos que ya visitamos.
    visitados = set()

    # Mientras haya caminos dentro de la pila,
    # continuamos buscando.
    while pila:

        # Sacamos el ÚLTIMO camino de la pila.
        # pop() sin ningún número elimina el último elemento.
        #
        # Esto hace que funcione como LIFO:
        # Last In, First Out
        # (el último en entrar es el primero en salir).
        camino = pila.pop()

        # Obtenemos el último nodo del camino actual.
        nodo_actual = camino[-1]

        # Comprobamos si llegamos al objetivo.
        if nodo_actual == objetivo:

            # Si llegamos, devolvemos el camino completo.
            return camino

        # Comprobamos si el nodo todavía no ha sido visitado.
        if nodo_actual not in visitados:

            # Marcamos el nodo como visitado.
            visitados.add(nodo_actual)

            # Recorremos todos los vecinos del nodo actual.
            for vecino in grafo[nodo_actual]:

                # Creamos un nuevo camino agregando
                # el vecino al camino actual.
                nuevo_camino = camino + [vecino]

                # Agregamos el nuevo camino a la PILA.
                # Como es una pila, se procesará posteriormente
                # siguiendo el principio LIFO.
                pila.append(nuevo_camino)

    # Si se vacía la pila y no encontramos el objetivo,
    # significa que no existe un camino.
    return None


# Ejecutamos DFS utilizando:
# mapa       -> nuestro grafo
# Almacen    -> punto de inicio
# ZonaNorte  -> objetivo
resultado_dfs = dfs(mapa, "Almacen", "ZonaNorte")

# Mostramos el camino encontrado.
print("Camino encontrado (DFS):", resultado_dfs)
```
